```python
import os 
import json
import requests
```

1st request: n. of events between 8/10/2026 and 8/11/2026

```python
URL_count = "https://earthquake.usgs.gov/fdsnws/event/1/count?starttime=2026-10-08&endtime=2026-11-08"

risposta = requests.get(URL_count, timeout = 30)
risposta.raise_for_status()
print(f"Request status: {risposta.status_code}")
print(f"Number of events from 8th October to 8th November: {risposta.text}")
```

2nd request: get 8/10/2026 - 9/10/2026 data

```python
URL = "https://earthquake.usgs.gov/fdsnws/event/1/query?format=geojson&starttime=2026-10-08&endtime=2026-11-08"
risposta = requests.get(URL, timeout = 30)
risposta.raise_for_status()
print(f"Request status: {risposta.status_code}")
```

Create "raw" directory and save data in .geojson format

```python
os.makedirs("raw", exist_ok=True)

with open("raw/0810_0811.geojson", "w", encoding="utf-8") as f:
    f.write(risposta.text)
```

Load the .geojson file and perform some checks:

- type should be dictionary
- show keys
- length of "features" should match the output of the 1st request (count)

```python
with open("raw/0810_0811.geojson", encoding="utf-8") as f:
    dati = json.load(f)
```

```python
print(type(dati))
print(dati.keys())
print(f"\nNumber of events: {len(dati["features"])}\n")

primo = dati["features"][0]
print(f"First event:\n {json.dumps(primo,indent = 2)}")
```

Function to extract and save data:
- Set "raw" as the directory to store data
- Set the URL without parameters
- Set a dictionary of parameters: the user can choose the format, the start date, the minimum magnitude and the (number of events) limit

The function creates the "raw" directory (if it does not exisist) and writes a geojson file in it.

```python
raw_dir = "raw"
URL = "https://earthquake.usgs.gov/fdsnws/event/1/query"

params = {"format": "geojson", "starttime": "2026-10-08", "minmagnitude": 2.5, "limit": 20000}


def extract_and_save(params, file_name):

    response = requests.get(URL, params=params, timeout=30)
    response.raise_for_status()

    os.makedirs(raw_dir, exist_ok=True)
    path_to_file = os.path.join(raw_dir, file_name)
    with open(path_to_file, "w", encoding = "utf-8") as f:
        f.write(response.text)

    return response.text
```

```python
extract_and_save(params, "try.geojson")
```
